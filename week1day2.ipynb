```python
#ЗАДАНИЕ 1 Написать декоратор `@timer` (замеряет время выполнения функции)

import time
from functools import wraps

def timer(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"Функция '{func.__name__}' выполнилась за {duration:.6f} секунд.")
        return result
    return wrapper
```

```python
def sum_of_two_numbers(a, b):
    return a+b
```

```python
def factorial(n):
    result = 1
    for i in range(1, n+1):
        result *= i 
    return result
```

```python
#ЗАДАНИЕ 2 Написать генератор чисел Фибоначчи через `yield`

def fibonacci_generator(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a+b
```

```python
fib = fibonacci_generator(3)
```

```python
list(fib)
```
