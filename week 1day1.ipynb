```python
# ЗАДАЧА 11 Даны два списка одинаковой длины: keys = ['a', 'b', 'c'] и values = [1, 2, 3]. Используя list comprehension (или dict comprehension),
# создайте из них один словарь { 'a': 1, ... }.

keys = ['a', 'b', 'c']
values = [1, 2, 3]

result = {k: v for k, v in zip(keys, values)}
```

```python
# ЗАДАЧА 12 Выпрямление матрицы (Flatten)
# Дан двумерный список (матрица) вида [[1, 2, 3], [4, 5], [6, 7, 8]]. Превратите его в один плоский список [1, 2, 3, 4, 5, 6, 7, 8].

matrix = [[1, 2, 3], [4, 5], [6, 7, 8]]
result = [x for sublist in matrix for x in sublist]
```

```python
result
```

```python
# ЗАДАЧА 13 Удаление гласных из слов
# Дан список слов. Напишите генератор, который вернет этот же список слов, но из каждого слова будут удалены все гласные буквы 
# (как английские, так и русские на ваш выбор).

words = ['pig', 'yogurt', 'toss', 'hall', 'oak', 'nose', 'ant', 'even', 'ill']
vowels = set('aeiouyAEIOUYауоыиэяюёеАУОЫИЭЯЮЁЕ')
result = [''.join([char for char in word if char not in vowels]) for word in words]
```

```python
result
```

```python
# ЗАДАЧА 14 Поиск общих делителей
#  Даны два списка чисел. Сгенерируйте третий список, в который попадут только те числа из первого списка, 
# которые одновременно присутствуют и во втором списке, причем они должны быть четными.

list_1 = [1, 2, 3, 4, 5]
list_2 = [2, 6, 4, 8, 9]
result = [x for x in list_1 if x%2==0 and x in list_2]
```

```python
result
```

```python
# ЗАДАЧА 15 Фильтрация матрицы
# Дан список списков с числами. Напишите вложенный генератор списков, который оставит структуру матрицы, 
# но внутри каждой строки отфильтрует и сохранит только четные числа, возведенные в квадрат

matrix = [[1, 2, 3], [4, 5], [6, 7, 8]]
result = [ [x**2 for x in sublist if x%2==0] for sublist in matrix]
```

```python
#ЗАДАНИЕ 2.0 Написать функцию `is_palindrome(s)`

def is_palindrome(s):
    s_lower = s.lower()
    return s_lower == s_lower[::-1]
```

```python
is_palindrome('Анна')
```

```python
is_palindrome('Даня')
```

```python
#ЗАДАНИЕ 3.0 Написать функцию `factorial(n)` (рекурсия + цикл)

def factorial(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result
```

```python
factorial(4)
```
